```python
from pathlib import Path
import json

# Project root
project_root = Path.cwd()

# Report generated by 04_Insight_Generation.ipynb
report_path = project_root / "reports" / "business_insights.json"

print("Project root:", project_root)
print("Report path:", report_path)
print("Exists:", report_path.exists())

# Load the JSON report
with open(report_path, "r", encoding="utf-8") as f:
    business_insights = json.load(f)

print("\nReport loaded successfully.")
print("Sections:", list(business_insights.keys()))
```

```python
# Extract sections from the generated business insights

overall = business_insights["overall"]
top_risks = business_insights["top_risks"]
top_product_risks = business_insights["top_product_risks"]
recommendations = business_insights["recommendations"]
product_recommendations = business_insights["product_recommendations"]

print("Overall KPIs:")
print(overall)

print("\nTop business risks:", len(top_risks))
print("Top product risks:", len(top_product_risks))
print("Business recommendations:", len(recommendations))
print("Product recommendations:", len(product_recommendations))
```

```python
# Build the executive summary

executive_summary = {
    "title": "Business Performance Executive Summary",
    "period": "2019–2020",
    "total_sales": overall["total_sales"],
    "total_profit": overall["total_profit"],
    "profit_margin": overall["profit_margin"],
    "return_rate": overall["return_rate"],
    "critical_risks": overall["critical_risk_count"],
    "warning_risks": overall["warning_risk_count"],
    "key_findings": [
        f"Total sales reached ${overall['total_sales']:,.2f}.",
        f"Total profit reached ${overall['total_profit']:,.2f}.",
        f"Overall profit margin was {overall['profit_margin']:.2f}%.",
        f"Return rate was {overall['return_rate']:.2f}%.",
        f"There are {overall['critical_risk_count']} critical business risks requiring attention.",
        f"There are {overall['warning_risk_count']} additional warning-level risks."
    ]
}

print("Executive summary created.\n")

for finding in executive_summary["key_findings"]:
    print("•", finding)
```

```python
# Build the key business risks section

key_business_risks = []

for risk in top_risks:
    key_business_risks.append({
        "area": risk["area"],
        "severity": risk["severity"],
        "message": risk["message"]
    })

print("Key Business Risks\n")

for i, risk in enumerate(key_business_risks, start=1):
    print(f"{i}. [{risk['severity'].upper()}] {risk['area']}")
    print(f"   {risk['message']}")
```

```python
# Build the critical product risks section

critical_product_risks = [
    risk for risk in top_product_risks
    if risk["severity"] in ["critical", "high"]
]

print("Critical Product Risks\n")

for i, risk in enumerate(critical_product_risks, start=1):
    print(f"{i}. [{risk['severity'].upper()}] {risk['product']}")
    print(f"   Risk Score: {risk['risk_score']:,.2f}")
    print(f"   {risk['message']}")
    print()
```

```python
# Build the business recommendations section

business_recommendations = []

for rec in recommendations:
    business_recommendations.append({
        "area": rec["area"],
        "severity": rec["severity"],
        "risk": rec["risk"],
        "recommendation": rec["recommendation"]
    })

print("Business Recommendations\n")

for i, rec in enumerate(business_recommendations, start=1):
    print(f"{i}. [{rec['severity'].upper()}] {rec['area']}")
    print(f"   Risk: {rec['risk']}")
    print(f"   Action: {rec['recommendation']}")
    print()
```

```python
# Build the product recommendations section

final_product_recommendations = []

for rec in product_recommendations:
    final_product_recommendations.append({
        "product": rec["product"],
        "severity": rec["severity"],
        "risk_score": rec["risk_score"],
        "recommendation": rec["recommendation"]
    })

print("Product Recommendations\n")

for i, rec in enumerate(final_product_recommendations, start=1):
    print(f"{i}. [{rec['severity'].upper()}] {rec['product']}")
    print(f"   Risk Score: {rec['risk_score']:,.2f}")
    print(f"   Action: {rec['recommendation']}")
    print()
```

```python
# Assemble the complete final report

final_report = {
    "report_title": "AI-Generated Business Insights Report",
    "period": "2019–2020",
    
    "executive_summary": executive_summary,
    
    "key_business_risks": key_business_risks,
    
    "critical_product_risks": critical_product_risks,
    
    "business_recommendations": business_recommendations,
    
    "product_recommendations": final_product_recommendations
}

print("Final report assembled successfully.")
print("Sections:")
for section in final_report:
    print("•", section)
```

```python
# Save the final report to the project reports folder

reports_dir = project_root / "reports"
reports_dir.mkdir(parents=True, exist_ok=True)

final_report_path = reports_dir / "final_business_report.json"

with open(final_report_path, "w", encoding="utf-8") as f:
    json.dump(final_report, f, indent=2, ensure_ascii=False)

print("Final report saved successfully.")
print("Path:", final_report_path)
print("Exists:", final_report_path.exists())
print("Size:", final_report_path.stat().st_size, "bytes")
```

```python
# Create a human-readable Markdown business report

md_path = reports_dir / "Final_Business_Report.md"

with open(md_path, "w", encoding="utf-8") as f:

    f.write("# AI-Generated Business Insights Report\n\n")
    f.write("**Analysis Period:** 2019–2020\n\n")
    f.write("---\n\n")

    # Executive Summary
    f.write("## 1. Executive Summary\n\n")

    f.write(f"- **Total Sales:** ${overall['total_sales']:,.2f}\n")
    f.write(f"- **Total Profit:** ${overall['total_profit']:,.2f}\n")
    f.write(f"- **Profit Margin:** {overall['profit_margin']:.2f}%\n")
    f.write(f"- **Return Rate:** {overall['return_rate']:.2f}%\n")
    f.write(f"- **Critical Risks:** {overall['critical_risk_count']}\n")
    f.write(f"- **Warning Risks:** {overall['warning_risk_count']}\n\n")

    f.write("### Key Findings\n\n")

    for finding in executive_summary["key_findings"]:
        f.write(f"- {finding}\n")

    f.write("\n---\n\n")

    # Business Risks
    f.write("## 2. Key Business Risks\n\n")

    for i, risk in enumerate(key_business_risks, 1):
        f.write(
            f"### {i}. {risk['area']} — "
            f"{risk['severity'].upper()}\n\n"
        )
        f.write(f"{risk['message']}\n\n")

    f.write("---\n\n")

    # Product Risks
    f.write("## 3. Critical Product Risks\n\n")

    for i, risk in enumerate(critical_product_risks, 1):
        f.write(
            f"### {i}. {risk['product']} — "
            f"{risk['severity'].upper()}\n\n"
        )
        f.write(f"**Risk Score:** {risk['risk_score']:,.2f}\n\n")
        f.write(f"{risk['message']}\n\n")

    f.write("---\n\n")

    # Business Recommendations
    f.write("## 4. Business Recommendations\n\n")

    for i, rec in enumerate(business_recommendations, 1):
        f.write(
            f"### {i}. {rec['area']} — "
            f"{rec['severity'].upper()}\n\n"
        )
        f.write(f"**Risk:** {rec['risk']}\n\n")
        f.write(f"**Recommended Action:** {rec['recommendation']}\n\n")

    f.write("---\n\n")

    # Product Recommendations
    f.write("## 5. Product Recommendations\n\n")

    for i, rec in enumerate(final_product_recommendations, 1):
        f.write(
            f"### {i}. {rec['product']} — "
            f"{rec['severity'].upper()}\n\n"
        )
        f.write(f"**Risk Score:** {rec['risk_score']:,.2f}\n\n")
        f.write(f"**Recommended Action:** {rec['recommendation']}\n\n")

    f.write("---\n\n")
    f.write(
        "*Generated from the cleaned Superstore dataset using "
        "the AI Data Insights Generator pipeline.*\n"
    )

print("Markdown report created successfully.")
print("Path:", md_path)
print("Exists:", md_path.exists())
print("Size:", md_path.stat().st_size, "bytes")
```
