```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("SampleSuperstore.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df["Returns"].value_counts(dropna=False)
```

```python
df = df.drop(columns=["ind1", "ind2"])
```

```python
df.columns
```

```python
df.shape
```

```python
df["Returned"] = df["Returns"].fillna(0)
```

```python
df["Returned"].value_counts()
```

```python
df[["Returns", "Returned"]].head(10)
```

```python
df.duplicated().sum()
```

```python
df.columns
```

```python
df.rename(columns={"Row ID+O6G3A1:R6": "Row_ID"}, inplace=True)
```

```python
df.columns
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)
```

```python
df.columns
```

```python
df[["order_date", "ship_date"]].dtypes
```

```python
df["order_date"] = pd.to_datetime(df["order_date"], dayfirst=True)
df["ship_date"] = pd.to_datetime(df["ship_date"], dayfirst=True)
```

```python
df[["order_date", "ship_date"]].dtypes
```

```python
df[["order_date", "ship_date"]].head()
```

```python
df["shipping_days"] = (
    df["ship_date"] - df["order_date"]
).dt.days
```

```python
df[["order_date", "ship_date", "shipping_days"]].head()
```

```python
df["shipping_days"].describe()
```

```python
df[["sales", "quantity", "profit"]].describe()
```

### Profit Validation

The `profit` column contains both positive and negative values.
Negative profit represents transactions where the business incurred a loss
and is therefore considered valid business information.

Extreme minimum and maximum profit values were inspected and found to
correspond to actual product transactions. Therefore, these values were
retained rather than removed.
